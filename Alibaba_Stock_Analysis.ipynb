##Cleaning Data

```python
data_df["Date"] = pd.to_datetime(data_df["Date"], errors="coerce")
data_df.info()
```

##Exploratory Data Analysis

###Analisis Deskriptif

```python
data_df.describe()
```

###EDA

```python
data_df.agg({
    "Close": "sum",
    "High": "sum",
    "Low": "sum",
    "Volume": "mean"
})
```

```python
# Hitung Moving Averages
data_df["MA_50"] = data_df["Close"].rolling(window=50, min_periods=1).mean()
data_df["MA_200"] = data_df["Close"].rolling(window=200, min_periods=1).mean()

data_df[["Date", "Close", "MA_50", "MA_200"]]
```

```python
ohlc = data_df.set_index("Date").resample("M").agg({
    "Open": "first",
    "High": "max",
    "Low": "min",
    "Close": "last"
})
ohlc
```

###Market Dynamics

```python
#Daily Return
data_df['Daily Return'] = data_df['Close'].pct_change()
data_df['Daily Return']
```

```python
# Volality
data_df["Volatility"] = data_df["Daily Return"].rolling(window=30).std()
data_df["Volatility"]
```

```python
data_df.set_index("Date").resample("M").agg({
    "Volume": "mean"
})
```
